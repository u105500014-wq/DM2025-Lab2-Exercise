# **Competition Code**

## 1. Preprocessing Steps

For effective emotion classification, I focused on preserving semantic richness while removing noise:

* **Text Cleaning:** I implemented a `clean_text_optimized` function that removed **URLs** and **HTML tags** but deliberately **preserved emojis and punctuation marks** (e.g., "!", "?"). These elements often carry strong emotional signals crucial for this task.
* **Tokenization:** I utilized pre-trained tokenizers corresponding to each model architecture (e.g., `microsoft/deberta-v3-large` and `xlm-roberta-large`).
* **Label Encoding:** Text labels (joy, anger, etc.) were converted into numerical IDs using `sklearn.preprocessing.LabelEncoder`.
* **Max Length:** The maximum sequence length was set to `128` with truncation and padding to optimize memory usage without losing critical information.

```python
# Environment: Google Colab Pro
import re
from transformers import AutoTokenizer
from sklearn.preprocessing import LabelEncoder

def clean_text_optimized(text: str) -> str:
    if not isinstance(text, str):
        return ""
    text = re.sub(r"http\S+|www\.\S+", "", text)
    text = re.sub(r"<.*?>", "", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

model_checkpoint = "microsoft/deberta-v3-large"
tokenizer = AutoTokenizer.from_pretrained(model_checkpoint)

def preprocess_function(examples):
    return tokenizer(
        examples["text"], 
        truncation=True, 
        padding="max_length", 
        max_length=128
    )

label_encoder = LabelEncoder()
```

## 2. Feature Engineering Steps

To enhance model performance and robustness, I applied the following techniques:

* **Handling Class Imbalance:**
    * I calculated `class_weights` using the `'balanced'` strategy from `sklearn`.
    * These weights were integrated into a custom `WeightedTrainer`, adjusting the `CrossEntropyLoss` to penalize errors on minority classes more heavily.
* **Pseudo-Labeling (Self-Distillation):**
    * This was the key feature engineering step for boosting the score.
    * I used my best-performing model (Score: 0.7036) to generate predictions for the test set.
    * These high-confidence predictions were then combined with the original training data to form a "Super Train Set" (approx. 64k samples).
    * Retraining on this augmented dataset helped the model refine its decision boundaries and improve generalization.

```python
# Environment: Google Colab Pro
import numpy as np
import torch
from sklearn.utils.class_weight import compute_class_weight
from torch import nn
from transformers import Trainer

class WeightedTrainer(Trainer):
    def __init__(self, class_weights, *args, **kwargs):
        super().__init__(*args, **kwargs)
        self.class_weights = class_weights

    def compute_loss(self, model, inputs, return_outputs=False, num_items_in_batch=None):
        labels = inputs.get("labels")
        outputs = model(**inputs)
        logits = outputs.get("logits")
        loss_fct = nn.CrossEntropyLoss(weight=self.class_weights)
        loss = loss_fct(logits.view(-1, self.model.config.num_labels), labels.view(-1))
        return (loss, outputs) if return_outputs else loss
```

## 3. Model Implementation Steps

My final submission, `Final_Submission_Last_Try.csv`, is the result of a **Voting Ensemble** strategy that combines the strengths of three distinct models to maximize diversity and accuracy.

* **Ensemble Architecture:**
    1.  **Model A (DeBERTa-v3-Large):** Trained with Seed 42. This model leverages DeBERTa's disentangled attention mechanism for deep semantic understanding.
    2.  **Model B (DeBERTa-v3-Large + Pseudo):** Trained with Seed 2024 using the augmented "Super Train Set." This model benefits from seeing the test distribution during training.
    3.  **Model C (XLM-RoBERTa-Large):** Trained as a "foreign expert." As a different architecture with a distinct pre-training objective, XLM-R provides a complementary perspective, helping to correct biases inherent to DeBERTa.

* **Training Configuration:**
    * **Optimizer:** AdamW with a carefully tuned learning rate of `7e-6` for large models.
    * **Batch Size Strategy:** I used `per_device_train_batch_size=2` combined with `gradient_accumulation_steps=16` to simulate an effective batch size of 32, overcoming GPU memory limitations.
    * **Precision:** `fp16=True` (mixed precision) was enabled to accelerate training and reduce memory usage.
    * **Evaluation Strategy:** I used `load_best_model_at_end=True` while monitoring `macro_f1` to ensure the final model checkpoint was the one with the best validation performance, preventing overfitting.

* **Training Dynamics & Model Selection:**
I monitored the training logs closely to select the best checkpoints. As shown in Figure, the model reached its peak F1 score at **Step 3600**, while the lowest validation loss was observed at **Step 4000**. I stopped training after Step 4200 as the validation loss started to increase, indicating the onset of overfitting.

![Figure: Training logs showing F1 peak and Overfitting point](./pics/training_log.png)

```python
# Environment: Google Colab Pro
from transformers import TrainingArguments
from collections import Counter

training_args = TrainingArguments(
    output_dir="./results",
    learning_rate=7e-6,
    per_device_train_batch_size=2,
    gradient_accumulation_steps=16,
    fp16=True,
    num_train_epochs=5,
    save_strategy="steps",
    load_best_model_at_end=True,
    metric_for_best_model="macro_f1",
    greater_is_better=True
)

def ensemble_vote(predictions_list):
    final_preds = []
    for i in range(len(predictions_list[0])):
        votes = [preds[i] for preds in predictions_list]
        winner = Counter(votes).most_common(1)[0][0]
        final_preds.append(winner)
    return final_preds
```
