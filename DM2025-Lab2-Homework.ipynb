# **Competition Code**

## 1. Preprocessing Steps

```python
### Add the code related to the preprocessing steps in cells inside this section
# 此組最高分的程式碼是在Colab上完成
# 讀取 tweets_DM.json
import pandas as pd
import json
from tqdm import tqdm
import re
import emoji

texts = []
ids = []
with open('tweets_DM.json', 'r') as f:
    for line in tqdm(f, desc="Loading tweets_DM.json"):
        record = json.loads(line)
        tweet = record['_source']['tweet']
        ids.append(tweet['tweet_id'])
        texts.append(tweet['message'])

df_posts = pd.DataFrame({'id': ids, 'text': texts})

# 讀取其他檔案並合併
data_identification = pd.read_csv('data_identification.csv')
emotion = pd.read_csv('emotion.csv')

train_df = data_identification[data_identification['identification'] == 'train'].merge(emotion, on='tweet_id')
test_df = data_identification[data_identification['identification'] == 'test']

train_df = train_df.rename(columns={'tweet_id': 'id'})
test_df = test_df.rename(columns={'tweet_id': 'id'})

train_df = train_df.merge(df_posts, on='id', how='left')
test_df = test_df.merge(df_posts, on='id', how='left')

# 文字清理
def clean_text(text):
    if not isinstance(text, str):
        return ""
    text = re.sub(r'http[s]?://[^\s]+', '', text)
    text = re.sub(r'@\w+', '', text)
    text = re.sub(r'#', ' ', text)
    text = emoji.demojize(text, language='en')
    text = re.sub(r':', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

train_df['text'] = train_df['text'].apply(clean_text)
test_df['text'] = test_df['text'].apply(clean_text)

print("Preprocessing completed")
```

## 2. Feature Engineering Steps

```python
### Add the code related to the feature engineering steps in cells inside this section
# 本次提交未進行額外特徵工程
# 多次實驗顯示加入文字長度、標點數量、情緒詞典分數等特徵均導致效能下降
# 最終決定僅使用清洗後的原始文字作為輸入

X_train = train_df['text'].tolist()
y_train = train_df['emotion'].tolist()
X_test = test_df['text'].tolist()
test_ids = test_df['id'].tolist()

print("Feature engineering: using cleaned text only")
```

## 3. Model Implementation Steps

```python
### Add the code related to the model implementation steps in cells inside this section
!pip install -q transformers datasets accelerate

from transformers import AutoTokenizer, AutoModelForSequenceClassification, Trainer, TrainingArguments
from datasets import Dataset
from sklearn.metrics import accuracy_score
import numpy as np

# 正確的標籤映射
label2id = {'anger': 0, 'disgust': 1, 'fear': 2, 'joy': 3, 'sadness': 4, 'surprise': 5}
id2label = {v: k for k, v in label2id.items()}

# 使用 bert-base-chinese
model_name = "bert-base-chinese"
tokenizer = AutoTokenizer.from_pretrained(model_name)

def tokenize_function(examples):
    return tokenizer(examples['text'], truncation=True, padding='max_length', max_length=128)

# 建立 dataset
train_dataset = Dataset.from_dict({'text': X_train, 'label': [label2id[l] for l in y_train]})
test_dataset = Dataset.from_dict({'text': X_test})

train_dataset = train_dataset.map(tokenize_function, batched=True)
test_dataset = test_dataset.map(tokenize_function, batched=True)

# 分割驗證集
train_val_split = train_dataset.train_test_split(test_size=0.1)
train_split = train_val_split['train']
val_split = train_val_split['test']

# 載入模型
model = AutoModelForSequenceClassification.from_pretrained(
    model_name, num_labels=6, id2label=id2label, label2id=label2id
)

# 訓練設定
training_args = TrainingArguments(
    output_dir='./results',
    num_train_epochs=4,
    per_device_train_batch_size=32,
    per_device_eval_batch_size=32,
    learning_rate=3e-5,
    warmup_steps=100,
    weight_decay=0.01,
    evaluation_strategy="epoch",
    save_strategy="epoch",
    load_best_model_at_end=True,
    metric_for_best_model="accuracy",
    logging_dir='./logs',
    report_to="none"
)

def compute_metrics(eval_pred):
    logits, labels = eval_pred
    predictions = np.argmax(logits, axis=-1)
    return {"accuracy": accuracy_score(labels, predictions)}

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_split,
    eval_dataset=val_split,
    compute_metrics=compute_metrics
)

# 訓練與預測
trainer.train()
predictions = trainer.predict(test_dataset)
pred_labels = np.argmax(predictions.predictions, axis=1)
pred_emotions = [id2label[idx] for idx in pred_labels]

# 產生 submission 並對齊官方順序
submission = pd.DataFrame({'id': test_ids, 'emotion': pred_emotions})
sample_sub = pd.read_csv('sampleSubmission.csv')
final_submission = sample_sub[['id']].merge(submission, on='id', how='left')
final_submission.to_csv('submission.csv', index=False)

print("Model training and prediction completed")
print("submission.csv has been generated")
```

```python
# ================================
# 2. Feature Engineering Steps
# ================================

from sklearn.feature_extraction.text import TfidfVectorizer

# TF-IDF 特徵（進階設定）
tfidf = TfidfVectorizer(
    max_features=20000,
    ngram_range=(1, 2),     # unigram + bigram
    min_df=2,
    max_df=0.95,
    stop_words='english',
    sublinear_tf=True       # 對高頻詞降權
)

X_train = tfidf.fit_transform(train_df['text_clean'])
X_test = tfidf.transform(test_df['text_clean'])
y_train = train_df['emotion']

print(f"詞彙量: {len(tfidf.vocabulary_)}")
print(f"訓練特徵形狀: {X_train.shape}")
```

```python
# ================================
# 3. Model Implementation Steps
# ================================

from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report
import joblib

# 訓練模型
model = MultinomialNB(alpha=0.05)  # 更小 alpha 提升泛化
model.fit(X_train, y_train)

# 訓練集評估
pred_train = model.predict(X_train)
acc = accuracy_score(y_train, pred_train)
print(f"\n訓練準確率: {acc:.4f}")
print("分類報告:")
print(classification_report(y_train, pred_train))

# 預測測試集
pred_test = model.predict(X_test)

# 產生 submission.csv
submission = pd.DataFrame({
    'id': test_df['id'],
    'emotion': pred_test
})

# 對齊 samplesubmission.csv 的 id 順序
sample_sub = pd.read_csv('samplesubmission.csv')
submission = submission.set_index('id').reindex(sample_sub['id']).reset_index()

# 儲存到當前資料夾
submission.to_csv('submission.csv', index=False)

# 儲存模型與特徵器（都在這個資料夾）
joblib.dump(model, 'nb_model.pkl')
joblib.dump(tfidf, 'tfidf_vectorizer.pkl')

print("\nsubmission.csv 已成功產生！")
print("模型與特徵器已儲存！")
submission.head(10)
```

```python
# use this code in Kaggle Notebook
# === CPU 版：XGBoost + TF-IDF（5 分鐘訓練）===
# === 最終成功版：test_df 不 merge emotion ===
import pandas as pd
import json
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.utils.class_weight import compute_sample_weight
from xgboost import XGBClassifier
from sklearn.preprocessing import LabelEncoder
import re

# 1. 路徑
data_path = '/kaggle/input/dm-lab-2-private-competition'

# 2. 解析 JSON
with open(f'{data_path}/final_posts.json') as f:
    raw_data = json.load(f)

posts = []
for item in raw_data:
    root = item.get('root', {})
    post = root.get('_source', {}).get('post', {})
    posts.append({
        'id': str(post.get('post_id')),
        'text': post.get('text', '')
    })

df_posts = pd.DataFrame(posts)

# 3. 讀取其他
df_split = pd.read_csv(f'{data_path}/data_identification.csv')
df_split['id'] = df_split['id'].astype(str)

df_emotion = pd.read_csv(f'{data_path}/emotion.csv')
df_emotion['id'] = df_emotion['id'].astype(str)

# 4. 合併
# train: merge 所有
train_df = df_posts.merge(df_split[df_split['split'] == 'train'], on='id').merge(df_emotion, on='id')

# test: 只 merge split，不 merge emotion！
test_df = df_posts.merge(df_split[df_split['split'] == 'test'], on='id')

print(f"訓練資料: {len(train_df)}, 測試資料: {len(test_df)}")

# 5. 清理
def clean(text):
    if not isinstance(text, str): return ""
    text = text.lower()
    text = re.sub(r'http[s]?://\S+', '', text)
    text = re.sub(r'@\w+', '', text)
    text = re.sub(r'#(\w+)', r'\1', text)
    text = re.sub(r'[^a-z0-9\s]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

train_df['text_clean'] = train_df['text'].apply(clean)
test_df['text_clean'] = test_df['text'].apply(clean)

train_df = train_df[train_df['text_clean'].str.len() > 0]
test_df = test_df[test_df['text_clean'].str.len() > 0]  # test 可以濾！

print(f"清理後訓練: {len(train_df)}, 測試: {len(test_df)}")

# 6. 標籤編碼
le = LabelEncoder()
train_df['label'] = le.fit_transform(train_df['emotion'])

# 7. TF-IDF
tfidf = TfidfVectorizer(max_features=30000, ngram_range=(1,3), min_df=2, stop_words='english', sublinear_tf=True)
X = tfidf.fit_transform(train_df['text_clean'])
y = train_df['label']
sample_weights = compute_sample_weight('balanced', y=y)

# 8. XGBoost + CPU
model = XGBClassifier(
    n_estimators=500,
    max_depth=7,
    learning_rate=0.1,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    n_jobs=-1,
    eval_metric='mlogloss'
)

print("開始訓練 XGBoost...")
model.fit(X, y, sample_weight=sample_weights)
print("訓練完成！")

# 9. 預測
X_test = tfidf.transform(test_df['text_clean'])
pred_nums = model.predict(X_test)
pred = le.inverse_transform(pred_nums)

# 10. 輸出
sample_sub = pd.read_csv(f'{data_path}/samplesubmission.csv')
submission = pd.DataFrame({'id': test_df['id'], 'emotion': pred})
submission = submission.set_index('id').reindex(sample_sub['id']).reset_index()
submission.to_csv('submission.csv', index=False)

print("submission.csv 已產生！快提交！")
submission.head()
```
